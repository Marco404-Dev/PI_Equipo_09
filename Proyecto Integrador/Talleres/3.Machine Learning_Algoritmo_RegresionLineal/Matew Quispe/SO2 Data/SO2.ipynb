```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

%matplotlib inline
```

```python
df = pd.read_csv('/content/SO2_daily_aqs_data_downloaded_2026-09-18 09_55_07.csv')

df.head()
```

```python
df.info()
```

```python
df.describe()
```

```python
df.columns
```

```python
plt.figure(figsize=(8,5))

sns.scatterplot(
    x='Daily Max SO2 Concentration',
    y='Daily AQI Value',
    data=df
)

plt.xlabel('Concentración máxima diaria de SO2')
plt.ylabel('AQI diario')
plt.title('SO2 vs AQI')

plt.show()
```

```python
X = df[['Daily Max SO2 Concentration']] # variable predictora
y = df['Daily AQI Value'] # variable que queremos predecir
```

```python
from sklearn.model_selection import train_test_split
```

```python
X_train, X_test, y_train, y_test = train_test_split(
    X,
    y,
    test_size=0.30, # 30% prueba y 70% entrenamiento
    random_state=101
)
```

```python
from sklearn.linear_model import LinearRegression
```

```python
lm = LinearRegression()
```

```python
lm.fit(X_train, y_train)
```

```python
pred = lm.predict(X_test)
pred[:10]
```

```python
print("Intercepto:", lm.intercept_)
print("Coeficiente:", lm.coef_)
```

```python
from sklearn.metrics import mean_absolute_error
from sklearn.metrics import mean_squared_error
from sklearn.metrics import r2_score
mae = mean_absolute_error(y_test, pred)

mse = mean_squared_error(y_test, pred)

rmse = np.sqrt(mse)

r2 = r2_score(y_test, pred)
```

```python
print("MAE:", mae)
print("MSE:", mse)
print("RMSE:", rmse)
print("R²:", r2)
```

```python
plt.figure(figsize=(8,5))

plt.scatter(y_test, pred)

plt.xlabel('AQI Real')
plt.ylabel('AQI Predicho')

plt.title('Valores reales vs valores predichos')

plt.show()
```

```python
plt.figure(figsize=(8,5))

plt.scatter(
    X_test['Daily Max SO2 Concentration'],
    y_test,
    label='Datos reales'
)

orden = np.argsort(X_test['Daily Max SO2 Concentration'])

plt.plot(
    X_test['Daily Max SO2 Concentration'].iloc[orden],
    pred[orden],
    color='red',
    label='Regresión'
)

plt.xlabel('Concentración máxima diaria de SO2')
plt.ylabel('AQI diario')

plt.title('Regresión lineal SO2 - AQI')

plt.legend()
plt.show()
```

```python
residuos = y_test - pred
plt.figure(figsize=(8,5))

sns.histplot(residuos, kde=True)

plt.xlabel('Error')
plt.title('Distribución de los residuos')

plt.show()
```

```python
from sklearn.model_selection import KFold
from sklearn.model_selection import cross_val_score
cv = KFold(
    n_splits=5,
    shuffle=True,
    random_state=42
)
```

```python
scores = cross_val_score(
    lm,
    X,
    y,
    cv=cv,
    scoring='r2'
)

scores
```

```python
print("R² de cada fold:")
print(scores)

print("\nR² promedio:")
print(scores.mean())

print("\nDesviación estándar:")
print(scores.std())
```

```python
plt.figure()

plt.boxplot(scores)

plt.title('R² - Validación cruzada')
plt.ylabel('R²')

plt.show()
```
